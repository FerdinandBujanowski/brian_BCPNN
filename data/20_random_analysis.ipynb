# Data Processing

## Loading Files

```python
def string_to_tuple(tuple_string):
    return (int(tuple_string[0]), int(tuple_string[1]))

def pattern_string_to_tuple_list(pattern_string):
    return sorted([string_to_tuple(s) for s in [c.replace('[', '').replace(';','').replace(']','').replace(',','') for c in pattern_string.split(' ')]])

def patterns_from_txt(filepath) -> list[list[tuple[int,int]]]:
    patterns = []
    with open(filepath, 'r') as file:
        lines = file.readlines()
        for line in lines:
            patterns.append(pattern_string_to_tuple_list(line))
    return patterns
```

```python
def get_mc_score(pattern_list, mc):
    return sum([1 if mc in p else 0 for p in pattern_list])

def get_mc_nonlin_score(pattern_list, mc):
    return sum([1 if mc in p else 0 for p in pattern_list])**2

def get_pattern_score(pattern_list, p_i):
    return sum([get_mc_score(pattern_list, mc) for mc in pattern_list[p_i]])

def get_pattern_nonlin_score(pattern_list, p_i):
    return np.prod([get_mc_score(pattern_list, mc) for mc in pattern_list[p_i]])
    # return sum([get_mc_nonlin_score(pattern_list, mc) for mc in pattern_list[p_i]])
```

Load all stats.csv files into their respective dataframes, load pattern int tuple lists

```python
CURRENT_SERIES = 'A'
ALL_SERIES = ['A', 'B']
BATCHES = ['30_1', '15_1', '15_2', '15_3', '15_4', '10_1', '10_2', '10_3', '5_1', '5_2']
```

```python
ind_dataframes: dict[dict[pd.DataFrame]] = {}
pattern_list:dict[list[tuple[int,int]]] = {}

for series in ALL_SERIES:
    CURRENT_PATH = f'tests_{series}/'
    pattern_list[series] = patterns_from_txt(f'{CURRENT_PATH}patterns_{series}.txt')
    ind_dataframes[series] = {}

    for batch in BATCHES:
        ind_dataframes[series][batch] = pd.read_csv(f'{CURRENT_PATH}stats_{batch}.csv')
```

Create new "processed" dataframes 

```python
for series in ALL_SERIES:
    for batch in BATCHES:
        ind_dataframes[series][batch]['original_pattern'] = [pattern_string_to_tuple_list(s) for s in ind_dataframes[series][batch]['original_pattern']]
        ind_dataframes[series][batch]['OP_i'] = [pattern_list[series].index(s) for s in ind_dataframes[series][batch]['original_pattern']]

        # turn True/False into 1 / 0
        ind_dataframes[series][batch]['reconstr_success'] = [1 if r else 0 for r in ind_dataframes[series][batch]['reconstr_success']]
        ind_dataframes[series][batch]['others_silent'] = [1 if r else 0 for r in ind_dataframes[series][batch]['others_silent']]
```

# Data Analysis

for batches 1, 2, 5: plot success rate for each pattern in function of index

```python
success_dicts: dict[dict[dict[int,float]]] = {}
interference_dicts: dict[dict[dict[int,float]]] = {}

for series in ALL_SERIES:
    success_dicts[series] = {}
    interference_dicts[series] = {}

    for batch in BATCHES:
        success_dicts[series][batch] = dict()
        interference_dicts[series][batch] = dict()
        for i_pattern in range(20):
            current_slice_df = ind_dataframes[series][batch][ind_dataframes[series][batch]['OP_i'] == i_pattern]
            success_dicts[series][batch][i_pattern] = len(current_slice_df[current_slice_df['reconstr_success'] == True]) / len(current_slice_df)
            if 'others_silent' in current_slice_df:
                interference_dicts[series][batch][i_pattern] = len(current_slice_df[current_slice_df['others_silent'] == False]) / len(current_slice_df)
```

```python
vir = plt.colormaps['viridis']

def plot_pattern_success(series, batches):
    for i, b in enumerate(batches):
        plt.plot(range(20), [success_dicts[series][b][i_p] for i_p in range(20)], label=f'Batch {b}', marker='x', alpha=0.7, c=vir(i/len(batches)))
        # if len(interference_dicts[b]) > 0:
        #     plt.plot(range(20), [interference_dicts[b][i_p] for i_p in range(20)], marker='o', alpha=0.4, c=vir(i/len(batches)))
    plt.xlabel('Pattern index (/ training order)')
    plt.xticks(np.arange(0, 20, 1))
    plt.yticks(np.arange(0, 1.1, 0.1))
    plt.ylabel('Reconstruction success (%)')
    plt.legend()
    plt.show()
```

for each batch: print batch total success rate

```python
def binom_std(p, n):
    return np.sqrt(n*p*(1-p))
```

```python
def print_rates(series):

    for batch in BATCHES:
        success_count = sum(ind_dataframes[series][batch]['reconstr_success'])
        interf_count = total_count - sum(ind_dataframes[series][batch]['others_silent'])
        total_count = len(ind_dataframes[series][batch])

        total_success_rate = success_count / total_count
        success_std = binom_std(total_success_rate, total_count) / total_count
        
        print(f'Average reconstruction success rate for batch {batch} = {round(total_success_rate*100,2)}% +/- {round(success_std*100,2)}% ({success_count}/{total_count})')
        total_interf_rate = interf_count / total_count
        interf_std = binom_std(total_interf_rate, total_count) / total_count
        print(f'Average interference rate for batch {batch} = {round(total_interf_rate*100,2)}+/-{round(interf_std*100,2)} ({interf_count}/{total_count})')
```

# Statistical Tests

perform t test on success or interference of two batches

```python
from scipy.stats import ttest_ind

def perform_t_test(series_1, series_2, batch_1, batch_2, reconstr=True):
    col_str = 'reconstr_success' if reconstr else 'others_silent'
    [dist_1, dist_2] = [ind_dataframes[s][b][col_str] for s, b in zip([series_1, series_2], [batch_1, batch_2])]

    return ttest_ind(dist_1, dist_2)
```

```python
def print_t_tests(series, reconstr):
    for i in range(len(BATCHES)):
        batch_i = BATCHES[i]
        for j in range(i+1, len(BATCHES)):
            batch_j = BATCHES[j]
            test_result = perform_t_test(series, series, batch_i, batch_j, reconstr)
            print(f'P-value for T-Test between batch {batch_i} and batch {batch_j} = {test_result.pvalue}')
```

```python
print_t_tests('A', True)
```

```python
def batch_wise_series_t_test(reconstr):
    for i in range(len(BATCHES)):
        batch_i = BATCHES[i]
        test_result = perform_t_test('A', 'B', batch_i, batch_i, reconstr)
        print(f'P-value for T-Test between batch A.{batch_i} and batch B.{batch_i} = {test_result.pvalue}')
```

```python
print('Reconstruction:')
batch_wise_series_t_test(True)

print('\nInterference:')
batch_wise_series_t_test(False)
```

```python
def plot_bar_stats(ax, series, batches, reconstr=True, ylabel=None, color='tab:blue', series_list=None):
    rates = []
    stds = []
    for i_b, batch in enumerate(batches):
        current_series = series if series_list is None else series_list[i_b]
        total_count = len(ind_dataframes[current_series][batch])
        
        if reconstr:
            success_count = sum(ind_dataframes[current_series][batch]['reconstr_success'])
            total_success_rate = success_count / total_count
            success_std = binom_std(total_success_rate, total_count) / total_count
            rates.append(total_success_rate*100)
            stds.append(success_std*100)
        else:
            interf_count = total_count - sum(ind_dataframes[current_series][batch]['others_silent'])
            total_interf_rate = interf_count / total_count
            interf_std = binom_std(total_interf_rate, total_count) / total_count
            rates.append(total_interf_rate*100)
            stds.append(interf_std*100)

    ax.set_ylim(0, 120)
    ax.set_yticks(np.arange(0, 120, 20))
    
    if ylabel is not None:
        ax.set_ylabel(ylabel)
    
    ax.bar([f'{series_list[i_b] + "." if series_list is not None else ""}{b[:-2]}.{b[-1:]}' for i_b, b in enumerate(batches)], rates, color=color)
    ax.errorbar(range(len(batches)), rates, stds, c='k', ls=' ')
```

```python
fig, axes = plt.subplots(2, 3, sharey=True)

plot_bar_stats(axes[0, 0], 'A', ['30_1', '15_1', '15_2', '15_3', '15_4'], ylabel='Reconstruction Success (%)')
plot_bar_stats(axes[0, 1], 'A', ['30_1', '10_1', '10_2', '10_3'])
plot_bar_stats(axes[0, 2], 'A', ['30_1', '5_1', '5_2'])

plot_bar_stats(axes[1, 0], 'A', ['30_1', '15_1', '15_2', '15_3', '15_4'], ylabel='Interference (%)', reconstr=False, color='tab:orange')
plot_bar_stats(axes[1, 1], 'A', ['30_1', '10_1', '10_2', '10_3'], reconstr=False, color='tab:orange')
plot_bar_stats(axes[1, 2], 'A', ['30_1', '5_1', '5_2'], reconstr=False, color='tab:orange')

# %matplotlib qt
%matplotlib inline
plt.show()
```
